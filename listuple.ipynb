```python
# Create a tuple of five fruits
fruits = ("apple", "banana", "kiwi", "orange", "papaya")

# Convert tuple to list
fruit_list = list(fruits)

# Add one new fruit
fruit_list.append("mango")

# Convert list back to tuple
fruits = tuple(fruit_list)

# Display the result
print("Updated tuple:", fruits)
```

```python
# Create a tuple containing marks of five students
marks = (85, 90, 78, 92, 88)

# Total number of marks
print("Total number of marks:", len(marks))

# Highest mark
print("Highest mark:", max(marks))

# Lowest mark
print("Lowest mark:", min(marks))

# Average mark
average = sum(marks) / len(marks)
print("Average mark:", average)
```
