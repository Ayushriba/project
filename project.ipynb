```python
# LogiNext Express - Delivery Management System

# Supported postal codes (fixed zone list)
supported_zones = ("560001", "560002", "560003", "560004", "560005")


# Package database using list of dictionaries
packages = [
    {
        "id": 101,
        "client": "Amazon",
        "destination": "560001",
        "weight": 2.5,
        "status": "Sorted"
    },

    {
        "id": 102,
        "client": "shein",
        "destination": "560009",
        "weight": 1.5,
        "status": "Sorted"
    },

    {
        "id": 103,
        "client": "temu",
        "destination": "560003",
        "weight": 3.0,
        "status": "Sorted"
    }
]


# Delivery stack
delivery_stack = []


while True:

    print("\n===== LogiNext Express =====")
    print("1. Display All Packages")
    print("2. Check Zones")
    print("3. View Loading Order")
    print("4. Deliver Package")
    print("5. Exit")


    option = input("Enter option: ")


    # Display package records
    if option == "1":

        print("\nPackage Details:")

        for package in packages:

            print("----------------")
            print("Package ID:", package["id"])
            print("Client:", package["client"])
            print("Destination:", package["destination"])
            print("Weight:", package["weight"], "kg")
            print("Status:", package["status"])



    # Verify zones
    elif option == "2":

        print("\nZone Verification:")

        for package in packages:

            if package["destination"] in supported_zones:

                print(package["id"], "Approved")

                delivery_stack.append(package)

                package["status"] = "Dispatched"

            else:

                print(package["id"],
                      "Rejected - Area Not Supported")



    # Show loading order
    elif option == "3":

        print("\nVan Loading Sequence:")

        if len(delivery_stack) == 0:

            print("No packages loaded")

        else:

            count = 1

            for package in delivery_stack:

                print(count,
                      "Package:",
                      package["id"],
                      "Destination:",
                      package["destination"])

                count += 1



    # Deliver package
    elif option == "4":

        if len(delivery_stack) > 0:

            delivered = delivery_stack.pop()

            delivered["status"] = "Delivered"

            print("\nDelivered:")
            print("Package ID:", delivered["id"])
            print("Client:", delivered["client"])

        else:

            print("No packages available")



    # Exit
    elif option == "5":

        print("Closing Logistics System...")
        break


    else:

        print("Invalid Selection")
```
