```python
# Create 3 student dictionaries
student1 = {
    "first_name": "Ayushri",
    "last_name": "Jadeja",
    "roll_number": 11,
    "course": "Computer Science"
}

student2 = {
    "first_name": "hardik",
    "last_name": "patel",
    "roll_number": 12,
    "course": "Information Technology"
}

student3 = {
    "first_name": "gargi",
    "last_name": "zala",
    "roll_number": 13,
    "course": "Data Science"
}

# Store all students in a list
students = [student1, student2, student3]

# Iterate through all students
for student in students:
    # Access and print last_name
    print("Last Name:", student["last_name"])

    # Remove last_name key using pop()
    student.pop("last_name")

# Display updated dictionaries
print("\nUpdated student dictionaries:")

for student in students:
    print(student)
```
