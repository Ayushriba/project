```python
postal_codes = {
    "M5V 2T6",
    "M5V 2T7",
    "M5V 2T8",
    "L5B 4M7",
    "L5B 4M8",
    "L5B 4M9",
    "N1G 5A1",
    "N1G 5A2"
}

parcel_store = {
    101: {
        "Customer": "Rita",
        "Address": "M5V 2T6",
        "Weight": 2,
        "Current Status": "Label Generated"
    },
    102: {
        "Customer": "Karan",
        "Address": "M5V 2T7",
        "Weight": 5,
        "Current Status": "Sorted"
    },
    103: {
        "Customer": "Meet",
        "Address": "L5B 4M7",
        "Weight": 1.5,
        "Current Status": "Out for Delivery"
    },
    104: {
        "Customer": "David",
        "Address": "L5B 4M8",
        "Weight": 7,
        "Current Status": "Delivered"
    },
    105: {
        "Customer": "Diya",
        "Address": "N1G 5A1",
        "Weight": 3,
        "Current Status": "Sorted"
    },
    106: {
        "Customer": "Ayushri",
        "Address": "N1G 5A2",
        "Weight": 4.5,
        "Current Status": "Label Generated"
    },
    107: {
        "Customer": "Hardik",
        "Address": "M5V 2T8",
        "Weight": 6,
        "Current Status": "Delivered"
    },
    108: {
        "Customer": "Gargi",
        "Address": "L5B 4M9",
        "Weight": 2.8,
        "Current Status": "Out for Delivery"
    }
}

dispatch_list = {}

drivers = [
    "Daniel",
    "Charlotte",
    "Michael",
    "Ava",
    "Ethan",
    "Grace"
]

next_id = max(parcel_store.keys()) + 1

running = True

while running:

    print("\n===================================")
    print("      FAST TRACK COURIER SYSTEM")
    print("===================================")
    print("1. Register New Parcel")
    print("2. Display All Parcels")
    print("3. Update Parcel Status")
    print("4. Dispatch Parcel")
    print("5. Search Parcel by Tracking ID")
    print("6. Check Serviceable Postal Code")
    print("7. View Delivery Drivers")
    print("8. Exit")
    print("===================================")

    try:
        option = int(input("Enter your choice (1-8): "))
    except ValueError:
        print(" Invalid input! Please enter a number from 1 to 8.")
        continue

    # Register Parcel
    if option == 1:

        customer = input("\nEnter Customer Name: ")

        while True:
            address = input("Enter Destination Postal Code: ").upper()

            if address in postal_codes:
                break

            print("This postal code is not serviceable. Please try again.")

        while True:
            try:
                weight = float(input("Enter Parcel Weight (kg): "))
                break
            except ValueError:
                print(" Please enter a valid weight.")

        parcel_store[next_id] = {
            "Customer": customer,
            "Address": address,
            "Weight": weight,
            "Current Status": "Label Generated"
        }

        print("\n Parcel Registered Successfully!")
        print("Tracking Number:", next_id)

        next_id += 1

    # Display Parcels
    elif option == 2:

        if len(parcel_store) == 0:
            print("\nNo parcels available.")

        else:
            print("\n========== PARCEL LIST ==========")

            for number in parcel_store:

                print("\nTracking ID:", number)

                for key in parcel_store[number]:
                    print(f"{key}: {parcel_store[number][key]}")

    # Update Status
    elif option == 3:

        number = int(input("\nEnter Tracking ID: "))

        if number not in parcel_store:
            print("Parcel not found.")
            continue

        print("\nSelect New Status")
        print("1. Sorted")
        print("2. Out for Delivery")
        print("3. Delivered")

        status = input("Enter your choice: ")

        if status == "1":
            parcel_store[number]["Current Status"] = "Sorted"

        elif status == "2":
            parcel_store[number]["Current Status"] = "Out for Delivery"

        elif status == "3":
            parcel_store[number]["Current Status"] = "Delivered"

        else:
            print(" Invalid status selection.")
            continue

        print("Parcel status updated successfully.")

    # Dispatch Parcel
    elif option == 4:

        number = int(input("\nEnter Tracking ID to Dispatch: "))

        if number in parcel_store:

            item = parcel_store.pop(number)

            item["Current Status"] = "Out for Delivery"
            item["Driver"] = drivers[number % len(drivers)]

            dispatch_list[number] = item

            print("\n Parcel dispatched successfully.")
            print("Assigned Driver:", item["Driver"])

        else:
            print("Tracking ID not found.")

    # Search Parcel
    elif option == 5:

        number = int(input("\nEnter Tracking ID to Search: "))

        if number in parcel_store:

            print("\nParcel Details")
            for key, value in parcel_store[number].items():
                print(f"{key}: {value}")

        elif number in dispatch_list:

            print("\nParcel Details")
            for key, value in dispatch_list[number].items():
                print(f"{key}: {value}")

        else:
            print("Parcel not found.")

    # Postal Code Check
    elif option == 6:

        area = input("\nEnter Postal Code to Check: ").upper()

        if area in postal_codes:
            print("Delivery service is available in this area.")

        else:
            print("Sorry, delivery service is not available for this postal code.")

    # Driver List
    elif option == 7:

        print("\nAvailable Delivery Drivers")
        print("----------------------------")

        index = 1
        while index <= len(drivers):
            print(f"{index}. {drivers[index - 1]}")
            index += 1

    # Exit
    elif option == 8:

        print("\nThank you for using Fast Track Courier System.")
        print("Goodbye!")
        running = False

    else:
        print("Invalid choice! Please select a number between 1 and 8.")
```
