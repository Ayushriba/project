```python
import numpy as np

# 5 different 3D NumPy arrays
arrays = [
    np.array([[[2, 5], [8, 11]],
              [[14, 17], [20, 23]]]),

    np.array([[[31, 34, 37],
               [40, 43, 46]]]),

    np.array([[[52], [55], [58]],
              [[61], [64], [67]],
              [[70], [73], [76]]]),

    np.array([[[81, 84],
               [87, 90],
               [93, 96]],
              [[99, 102],
               [105, 108],
               [111, 114]]]),

    np.array([[[121, 124],
               [127, 130]],
              [[133, 136],
               [139, 142]],
              [[145, 148],
               [151, 154]]])
]

# Display information for each array
for n, arr in enumerate(arrays, start=1):
    print(f"\n========== Array {n} ==========")

    # Display complete array
    print("Complete Array:")
    print(arr)

    # Number of dimensions
    print("Number of dimensions (ndim):", arr.ndim)

    # Access and print every element using indexing
    print("Elements:")
    for i in range(arr.shape[0]):
        for j in range(arr.shape[1]):
            for k in range(arr.shape[2]):
                print(f"arr[{i}][{j}][{k}] = {arr[i][j][k]}")
```
